```python
import heapq

def uniform_cost_search(graph, start, goals):
    """
    Implements the Uniform Cost Search algorithm to find the minimum cost path.

    :param graph: Dictionary where keys are nodes and values are lists of tuples (neighbor, cost).
    :param start: The starting node.
    :param goals: A set of goal states.
    :return: The path with the minimum cost to one of the goal states and its cost.
    """
    # priority queue: (cost, node, path)
    pq = [(0, start, [start])]  
    visited = {}

    while pq:
        #the pq will always keep the lowest cost node path first to pop
        cost, node, path = heapq.heappop(pq)

        if node in visited and visited[node] <= cost:
            continue
        visited[node] = cost

        if node in goals:  # if goal reached, return path and cost
            print(f"Minimum cost path: {path} with cost {cost}")
            return path, cost

        for neighbor, edge_cost in graph.get(node, []):
            new_cost = cost + edge_cost
            heapq.heappush(pq, (new_cost, neighbor, path + [neighbor]))

    print("No path found.")
    return None, float('inf')

```

```python
graph = {
    0: [(4, 3), (5, 9), (16, 1)], #[node number, cost going to that node from the current node]
    1: [],
    2: [(1, 2)],
    3: [(1, 8)],
    4: [(3, 8), (2,1), (5,2)],
    5: [(18, 2), (8, 8)],
    6: [(7, 2,), (3,2)],
    7: [],
    8: [(16, 8), (17,8)],
    16: [(17, 1)],
    17: [],
    18: [(6, 1)],
}

start = 0
goal_states = {7}

uniform_cost_search(graph, start, goal_states)
```
